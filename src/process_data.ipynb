# Khám phá và xử lý dữ liệu Global Superstore

Dùng các hàm có sẵn của pandas để xem dữ liệu, kiểm tra chất lượng và chuẩn hóa đơn giản.

- Đầu vào: `data/superstore.csv`.
- Đầu ra duy nhất: `data/processed/superstore_processed.csv`.
- Giữ nguyên `Row.ID`, `weeknum`, các mã và số liệu nguồn.
- Không tạo khóa mới, bảng chiều hoặc các tệp báo cáo phụ.

Nguồn: https://www.kaggle.com/datasets/fatihilhan/global-superstore-dataset

## 1. Đọc dữ liệu
Đường dẫn dưới đây dùng được khi mở notebook từ thư mục `src` hoặc thư mục gốc của dự án.

```python
from pathlib import Path
import pandas as pd

data_dir = Path("../data")
if not (data_dir / "superstore.csv").exists():
    data_dir = Path("data")

raw = pd.read_csv(data_dir / "superstore.csv", encoding="utf-8-sig")
print("Kích thước dữ liệu:", raw.shape)
raw.head()
```

## 2. Xem cấu trúc dữ liệu
Xem kiểu dữ liệu, giá trị thiếu và số giá trị phân biệt. Bảng thuộc tính có STT theo đúng thứ tự cột trong CSV.

```python
raw.info()

attributes = pd.DataFrame({
    "STT": range(1, len(raw.columns) + 1),
    "Thuộc tính": raw.columns,
    "Kiểu dữ liệu": raw.dtypes.astype(str).to_numpy(),
    "Số ô thiếu": raw.isna().sum().to_numpy(),
    "Số giá trị phân biệt": raw.nunique().to_numpy()
})
attributes
```

```python
raw.describe().T
```

## 3. Chuẩn hóa tên cột, văn bản và kiểu dữ liệu
Đổi tên cột thành chữ thường, thay dấu chấm bằng gạch dưới. `Market2` đổi thành `market_group`; cột đếm bản ghi được gọi là `record_count`.

Chỉ bỏ khoảng trắng thừa ở hai đầu văn bản. Chuyển ngày và các đại lượng số về kiểu phù hợp, không đổi giá trị hay tạo lại mã nguồn.

```python
df = raw.copy()

df.columns = df.columns.str.lower().str.replace(".", "_", regex=False)
df = df.rename(columns={"market2": "market_group", "记录数": "record_count"})

for col in df.select_dtypes(include=["object", "string"]).columns:
    df[col] = df[col].str.strip().str.normalize("NFC").replace("", pd.NA)

df["order_date"] = pd.to_datetime(df["order_date"], format="%Y-%m-%d %H:%M:%S.%f")
df["ship_date"] = pd.to_datetime(df["ship_date"], format="%Y-%m-%d %H:%M:%S.%f")

df = df.astype({
    "row_id": "int64", "quantity": "int64", "weeknum": "int64",
    "sales": "float64", "profit": "float64",
    "discount": "float64", "shipping_cost": "float64"
})
df.head()
```

## 4. Kiểm tra chất lượng
Kiểm tra ô thiếu, dòng trùng và một số điều kiện cơ bản. Lợi nhuận âm và doanh thu bằng 0 được thống kê riêng để lưu ý, không tự xem là lỗi hoặc loại bỏ.

```python
print("Số ô thiếu:", df.isna().sum().sum())
print("Số dòng trùng hoàn toàn:", df.duplicated().sum())
print("Số Row.ID bị lặp:", df["row_id"].duplicated().sum())
print("Số dòng lợi nhuận âm:", (df["profit"] < 0).sum())
print("Số dòng doanh thu bằng 0:", (df["sales"] == 0).sum())

checks = pd.Series({
    "Ngày gửi trước ngày đặt": (df["ship_date"] < df["order_date"]).sum(),
    "Số lượng không dương": (df["quantity"] <= 0).sum(),
    "Chiết khấu ngoài khoảng 0–1": (~df["discount"].between(0, 1)).sum(),
    "Doanh thu âm": (df["sales"] < 0).sum(),
    "Phí vận chuyển âm": (df["shipping_cost"] < 0).sum(),
    "Tuần nguồn ngoài khoảng 1–53": (~df["weeknum"].between(1, 53)).sum()
})
checks
```

## 5. Khảo sát mã định danh và cột phụ trợ
Dùng `groupby`, `nunique` và `drop_duplicates` để kiểm tra mã. Không gộp các dòng chỉ vì trùng mã đơn hoặc mã sản phẩm.

Các tổ hợp nhận diện chỉ dùng để khảo sát ở bước này. Không tạo hoặc xuất bảng chiều; phần thiết kế kho được thực hiện sau.

```python
product_names = df.groupby("product_id")["product_name"].nunique()
order_dates = df.groupby("order_id")["order_date"].nunique()
order_customers = df.groupby("order_id")["customer_id"].nunique()

orders = df[["order_id", "order_date", "customer_id"]].drop_duplicates()
products = df[["product_id", "product_name", "category", "sub_category"]].drop_duplicates()

print("Mã sản phẩm có nhiều tên:", (product_names > 1).sum())
print("Mã đơn có nhiều ngày đặt:", (order_dates > 1).sum())
print("Mã đơn có nhiều khách hàng:", (order_customers > 1).sum())
print("Số tổ hợp nhận diện đơn:", len(orders))
print("Số tổ hợp nhận diện sản phẩm:", len(products))

df[["market", "market_group"]].drop_duplicates().sort_values("market")
```

```python
print("Các giá trị của cột đếm:", df["record_count"].unique())
print("Năm nguồn khớp ngày đặt:", df["year"].eq(df["order_date"].dt.year).all())

iso_week = df["order_date"].dt.isocalendar().week
print("Số dòng tuần nguồn khác tuần ISO:", df["weeknum"].ne(iso_week).sum())
print("Vẫn giữ nguyên weeknum vì chưa rõ quy ước tuần của nguồn.")
```

## 6. Hoàn thiện dữ liệu
- Bỏ `record_count` vì toàn bộ giá trị bằng 1.
- Bỏ `year` vì trùng với năm của ngày đặt. Năm–quý–tháng sẽ được tổ chức trong `dim_date` ở bước thiết kế kho.
- Giữ `market_group`, `row_id`, `weeknum` và các thành phần nhận diện đơn, sản phẩm.
- Giữ `profit_margin` bằng lợi nhuận chia doanh thu; để trống khi doanh thu bằng 0.
- Không bổ sung cột khóa, cờ phân loại hay chênh lệch ngày.

Các kiểm tra ở trên cho thấy dữ liệu hiện tại không thiếu và không trùng hoàn toàn, nên không cần điền bù hoặc xóa dòng.

```python
df = df.drop(columns=["record_count", "year"])
df["profit_margin"] = df["profit"] / df["sales"].where(df["sales"] != 0)

print("Kích thước sau xử lý:", df.shape)
print("Số biên lợi nhuận không xác định:", df["profit_margin"].isna().sum())
df.head()
```

## 7. Đối chiếu số liệu và lưu kết quả
So sánh các tổng trước–sau. Xuất một CSV duy nhất, không ghi đè dữ liệu nguồn. File CSV không lưu kiểu dữ liệu, nên khi sử dụng lại cần chuyển hai cột ngày về kiểu ngày.

```python
comparison = pd.DataFrame({
    "Trước xử lý": raw[["Sales", "Profit", "Quantity", "Shipping.Cost"]].sum().to_numpy(),
    "Sau xử lý": df[["sales", "profit", "quantity", "shipping_cost"]].sum().to_numpy()
}, index=["Doanh thu", "Lợi nhuận", "Số lượng", "Phí vận chuyển"])
comparison["Chênh lệch"] = comparison["Sau xử lý"] - comparison["Trước xử lý"]
comparison
```

```python
output_dir = data_dir / "processed"
output_dir.mkdir(parents=True, exist_ok=True)
output_path = output_dir / "superstore_processed.csv"

df.to_csv(output_path, index=False, encoding="utf-8-sig", date_format="%Y-%m-%d")
print("Đã lưu:", output_path.resolve())
print(f"Kết quả: {len(df):,} dòng, {len(df.columns)} cột.")
```
